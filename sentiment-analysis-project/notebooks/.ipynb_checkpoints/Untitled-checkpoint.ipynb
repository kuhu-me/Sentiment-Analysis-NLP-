```python
data_set["Phrase"] = data_set["Phrase"].apply(lambda x: re.sub(r"[^\w\s]", "", x))
```

```python
data_set['Phrase']
```

```python
data_set["Tokens"] = data_set["Phrase"].apply(lambda x:word_tokenize(x))
```

```python
data_set['Tokens']
```

```python
nltk.download('stopwords')
stop_words=set(nltk.corpus.stopwords.words('english'))
stop_words
```

```python
data_set['Tokens']=data_set['Tokens'].apply(lambda tokens: [word for word in tokens if word.lower() not in stop_words])
```

```python
data_set['Tokens']
```

```python
nltk.download('wordnet')
```

```python
lem=WordNetLemmatizer()
data_set["Lemmatized_Tokens"] = data_set["Tokens"].apply(lambda tokens: [lem.lemmatize(word) for word in tokens])
```

```python
data_set["Lemmatized_Tokens"]
```

```python
data_set["Lemmitized_phrase"] = data_set["Lemmatized_Tokens"].apply(lambda tokens: " ".join(tokens))
```

```python
tfidf=TfidfVectorizer(ngram_range=(1, 2))
matrix = tfidf.fit_transform(data_set["Lemmitized_phrase"])
print(matrix)
```

```python
Y_train = data_set['Sentiment']               
X_train = matrix                 
X_test = matrix_test             
print(X_train,X_test)
```

```python
Y_train
```

```python
param_grid = {
    'C': [0.01, 0.1, 1, 10],  # Regularization strength
    'solver': ['lbfgs', 'liblinear', 'saga'],  # Solvers that work with small datasets
    'max_iter': [10000, 20000, 50000]
}
model=LogisticRegression(multi_class='auto',class_weight='balanced')

grid = GridSearchCV(estimator=model, param_grid=param_grid, cv=5, scoring='accuracy', verbose=2, n_jobs=-1)
```

```python
grid.fit(X_train, Y_train)
```

```python
best_model = grid.best_estimator_
predicted= best_model.predict(X_train)
accuracy = accuracy_score(Y_train, predicted)
print("Accuracy:", accuracy)
```

```python
print("Best parameters:", grid.best_params_)
```

```python
joblib.dump(best_model, 'sentiment_model.pkl')
```
